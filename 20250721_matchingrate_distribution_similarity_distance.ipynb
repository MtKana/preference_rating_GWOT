```python
# --- Configuration ---
n_iterations = 100
results_dir = "gwot_results_similarity_distance_averaged"
os.makedirs(results_dir, exist_ok=True)
n_eps = 15
eps_range = [0.04, 5]
epsilons = np.logspace(np.log10(eps_range[0]), np.log10(eps_range[1]), n_eps)

# --- Utility: Save iteration result ---
def save_iteration_result(index, result_data):
    with open(os.path.join(results_dir, f"iteration_{index:03}.pkl"), "wb") as f:
        pickle.dump(result_data, f)

# --- Utility: Check which iterations are done ---
def get_completed_iterations():
    return {
        int(f.split("_")[1].split(".")[0])
        for f in os.listdir(results_dir)
        if f.startswith("iteration_") and f.endswith(".pkl")
    }
```

```python
completed = get_completed_iterations()
for iteration in range(n_iterations):
    if iteration in completed:
        print(f"Iteration {iteration} already completed. Skipping.")
        continue

    print(f"\nStarting iteration {iteration}")

    # 1. Split subjects
    n_subjects = len(subject_matrices_similarity_distance)
    indices = list(range(n_subjects))
    random.shuffle(indices)

    mid = (n_subjects + 1) // 2
    group1_idx = indices[:mid]
    group2_idx = indices[mid:]

    group1 = [subject_matrices_similarity_distance[i] for i in group1_idx]
    group2 = [subject_matrices_similarity_distance[i] for i in group2_idx]

    # 2. Average matrices
    avg_matrix1 = np.mean(group1, axis=0)
    avg_matrix2 = np.mean(group2, axis=0)

    # 3. GWOT
    pair_of_preference = [avg_matrix1, avg_matrix2]
    OT_plan, gwds, matching_rates, min_gwd, best_matching_rate = GWD_and_plot(
        pair_of_preference[0], pair_of_preference[1], epsilons
    )

    # 4. Save result
    result_data = {
        "iteration": iteration,
        "group1_indices": group1_idx,
        "group2_indices": group2_idx,
        "OT_plan": OT_plan,
        "gwds": gwds,
        "matching_rates": matching_rates,
        "min_gwd": min_gwd,
        "best_matching_rate": best_matching_rate,
    }
    save_iteration_result(iteration, result_data)

    # 5. Plot heatmaps
    show_heatmaps(
        vmin_val=0,
        vmax_val=7,
        matrices=[avg_matrix1, avg_matrix2],
        titles=[
            f"Group 1 (avg {len(group1)})",
            f"Group 2 (avg {len(group2)})"
        ],
        nrows=1,
        ncols=2,
        cmap_name="viridis",
        cbar_label="Values",
        color_labels=unique_colours,
    )

    print(f"Completed iteration {iteration}, saved to file.")
```

```python
matching_rates = []

# --- Load all completed iteration files ---
for fname in os.listdir(results_dir):
    if fname.startswith("iteration_") and fname.endswith(".pkl"):
        with open(os.path.join(results_dir, fname), "rb") as f:
            data = pickle.load(f)
            if "best_matching_rate" in data:
                matching_rates.append(data["best_matching_rate"])

# --- Plot histogram ---
plt.figure(figsize=(8, 5))
plt.hist(matching_rates, bins=20, edgecolor='black', alpha=0.75)
plt.title("Distribution of Best Matching Rates across 100 Iterations")
plt.xlabel("Best Matching Rate")
plt.ylabel("Frequency")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import os
import pickle
import matplotlib.pyplot as plt

matching_rates = []
iteration_ids = []

# --- Load all completed iteration files ---
for fname in sorted(os.listdir(results_dir)):
    if fname.startswith("iteration_") and fname.endswith(".pkl"):
        with open(os.path.join(results_dir, fname), "rb") as f:
            data = pickle.load(f)
            if "best_matching_rate" in data:
                matching_rates.append(data["best_matching_rate"])
                iteration_ids.append(data["iteration"])

# --- Plot ---
plt.figure(figsize=(14, 6))
plt.plot(iteration_ids, matching_rates, marker='o', linestyle='-', color='blue')
plt.title("Exact Best Matching Rate per Iteration")
plt.xlabel("Iteration")
plt.ylabel("Best Matching Rate")
plt.grid(True)

# --- Annotate each point with the matching rate ---
for i, (x, y) in enumerate(zip(iteration_ids, matching_rates)):
    plt.text(x, y + 0.005, f"{y:.3f}", ha='center', va='bottom', fontsize=8, rotation=45)

plt.xticks(range(0, len(iteration_ids), max(1, len(iteration_ids)//20)))  # space x ticks nicely
plt.tight_layout()
plt.show()
```
