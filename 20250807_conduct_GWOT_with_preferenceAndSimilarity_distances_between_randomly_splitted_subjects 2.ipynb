```python
import numpy as np
import random
import matplotlib.pyplot as plt
from scipy.stats import ttest_rel
from utilityFunctions import (
    load_csv_to_matrix_batch,
    compute_color_similarity_distance_batch,
    compute_color_preference_distance_batch,
    show_heatmaps
)
from utilityFunctions_GWOT import GWD_and_plot
import pandas as pd
import os
import pickle
```

```python
# ---------------------------
# CONFIGURATION
# ---------------------------

n_iterations = 10
n_eps = 25
eps_range = [0.04, 2]
epsilons = np.logspace(np.log10(eps_range[0]), np.log10(eps_range[1]), n_eps)
n_subjects = 19

# Configuration for loading raw preference data
folder_path = 'raw_data/202411_data'
unique_colours = np.array(['#d2b700', '#db8b08', '#c7512c', '#c13547', '#a03663', '#753a7a', 
                           '#4b488e', '#005692', '#006a8b', '#007b75', '#008a52', '#9aa400'])
colour_index = {colour: idx for idx, colour in enumerate(unique_colours)}
matrix_size = len(unique_colours)

# Load raw similarity matrices for all 19 subjects
response_type = 'similarity'
subject_matrices_similarity_raw = load_csv_to_matrix_batch(
    folder_path,
    response_type,
    colour_index,
    matrix_size
)
response_type = 'preference'
subject_matrices_preference_raw = load_csv_to_matrix_batch(
    folder_path,
    response_type,
    colour_index,
    matrix_size
)

results_dir = "gwot_results_similarity_distance_preference_distance_between_subjects_half_averaged_ver2"
os.makedirs(results_dir, exist_ok=True)
```

```python
# ---------------------------
# INIT STORAGE
# ---------------------------

min_gwds_similarity = []
min_gwds_preference = []

# ---------------------------
# MAIN LOOP
# ---------------------------

for iteration in range(n_iterations):
    print(f"\nIteration {iteration + 1}/{n_iterations}")

    # 1. Random split
    indices = list(range(n_subjects))
    random.shuffle(indices)
    group1_idx = indices[:9]
    group2_idx = indices[9:]

    print(f"Group 1 indices: {group1_idx}, Group 2 indices: {group2_idx}") 

    group1_similarity_raw = [subject_matrices_similarity_raw[i] for i in group1_idx]
    group2_similarity_raw = [subject_matrices_similarity_raw[i] for i in group2_idx]
    group1_similarity_distance = compute_color_similarity_distance_batch(group1_similarity_raw)
    group2_similarity_distance = compute_color_similarity_distance_batch(group2_similarity_raw)
    avg_sim_1 = np.mean(group1_similarity_distance, axis=0)
    avg_sim_2 = np.mean(group2_similarity_distance, axis=0)

    group1_preference_raw = [subject_matrices_preference_raw[i] for i in group1_idx]
    group2_preference_raw = [subject_matrices_preference_raw[i] for i in group2_idx]
    group1_preference_distance = compute_color_preference_distance_batch(group1_preference_raw)
    group2_preference_distance = compute_color_preference_distance_batch(group2_preference_raw)
    avg_pref_1 = np.mean(group1_preference_distance, axis=0)
    avg_pref_2 = np.mean(group2_preference_distance, axis=0)

    show_heatmaps(
        vmin_val=0,
        vmax_val=7,
        matrices=[avg_sim_1, avg_sim_2],
        titles=["Group 1 Similarity", "Group 2 Similarity"],
        nrows=1,
        ncols=2,
        cmap_name="viridis",
        cbar_label="Similarity Distance",
        color_labels=None,
    )

    show_heatmaps(
        vmin_val=0,
        vmax_val=7,
        matrices=[avg_pref_1, avg_pref_2],
        titles=["Group 1 Preference", "Group 2 Preference"],
        nrows=1,
        ncols=2,
        cmap_name="viridis",
        cbar_label="Preference Distance",
        color_labels=None,
    )

    # 5. GWOT: Similarity
    sim_OT_plan, sim_gwds, sim_match_rates, sim_min_gwd, _ = GWD_and_plot(avg_sim_1, avg_sim_2, epsilons)

    # 6. GWOT: Preference
    pref_OT_plan, pref_gwds, pref_match_rates, pref_min_gwd, _ = GWD_and_plot(avg_pref_1, avg_pref_2, epsilons)

    # 7. Store minimal GWDs
    min_gwds_similarity.append(sim_min_gwd)
    min_gwds_preference.append(pref_min_gwd)

    sim = np.asarray(min_gwds_similarity, dtype=float)
    pref = np.asarray(min_gwds_preference, dtype=float)
    iterations = np.arange(1, len(sim) + 1)

    plt.figure(figsize=(10, 5))
    plt.plot(iterations, sim, marker='o', label='Similarity', alpha=0.8)
    plt.plot(iterations, pref, marker='s', label='Preference', alpha=0.8)

    plt.xlabel("Iteration")
    plt.ylabel("Minimal GWD")
    plt.title("Minimal GWDs at Each Iteration")
    plt.xticks(iterations)  # Show all iteration numbers on x-axis
    plt.legend()
    plt.grid(True, linestyle='--', alpha=0.6)
    plt.tight_layout()
    plt.show()

    # Get best matching rates (at index of min GWD)
    best_mr_similarity = sim_match_rates[np.argmin(sim_gwds)]
    best_mr_preference = pref_match_rates[np.argmin(pref_gwds)]

    # Prepare dictionary to save
    result_data = {
        "iteration": iteration,
        "min_gwd_similarity": sim_min_gwd,
        "min_gwd_preference": pref_min_gwd,
        "OT_plan_similarity": sim_OT_plan,
        "OT_plan_preference": pref_OT_plan,
        "best_matching_rate_similarity": best_mr_similarity,
        "best_matching_rate_preference": best_mr_preference
    }

    # Save as pickle
    save_path = os.path.join(results_dir, f"iteration_{iteration:03d}.pkl")
    with open(save_path, "wb") as f:
        pickle.dump(result_data, f)

    print(f"Saved iteration {iteration} results to {save_path}")
    # # 8. Plot GWD curves (only for first iteration)
    # if iteration == 0:
    #     plt.figure(figsize=(10, 4))

    #     plt.subplot(1, 2, 1)
    #     plt.plot(epsilons, sim_gwds, label="Similarity", marker='o')
    #     plt.plot(epsilons, pref_gwds, label="Preference", marker='s')
    #     plt.xlabel("Epsilon")
    #     plt.ylabel("GWD")
    #     plt.title("GWD vs Epsilon")
    #     plt.legend()

    #     plt.subplot(1, 2, 2)
    #     plt.imshow(sim_OT_plan, cmap="viridis", aspect='auto')
    #     plt.colorbar(label="Transport Mass")
    #     plt.title("OT Plan (Similarity)")
    #     plt.tight_layout()

    #     plt.show()
```

```python

results_dir = "gwot_results_similarity_distance_preference_distance_between_subjects_half_averaged_ver2"

min_gwds_similarity = []
min_gwds_preference = []
OT_plan_similarity = []
OT_plan_preference = []
matching_rate_similarity = []  # best per iteration
matching_rate_preference = []  # best per iteration
iteration_ids = []

for fname in sorted(os.listdir(results_dir)):
    if not (fname.startswith("iteration_") and fname.endswith(".pkl")):
        continue
    path = os.path.join(results_dir, fname)
    with open(path, "rb") as f:
        data = pickle.load(f)

    # Collect
    min_gwds_similarity.append(data["min_gwd_similarity"])
    min_gwds_preference.append(data["min_gwd_preference"])
    OT_plan_similarity.append(data["OT_plan_similarity"])
    OT_plan_preference.append(data["OT_plan_preference"])
    matching_rate_similarity.append(data["best_matching_rate_similarity"])
    matching_rate_preference.append(data["best_matching_rate_preference"])
    iteration_ids.append(data.get("iteration", len(iteration_ids)))  # fallback

# Convert to numpy arrays (handy for plotting/math)
min_gwds_similarity = np.asarray(min_gwds_similarity, dtype=float)
min_gwds_preference = np.asarray(min_gwds_preference, dtype=float)
matching_rate_similarity = np.asarray(matching_rate_similarity, dtype=float)
matching_rate_preference = np.asarray(matching_rate_preference, dtype=float)

# Use 1..N for x-axis (or use the recorded iteration_ids if you prefer)
iterations = np.arange(1, len(min_gwds_similarity) + 1)
```

```python
# ---------- Plot minimal GWDs with best matching rates labeled ----------
import matplotlib.pyplot as plt
import numpy as np

# Basic sanity check
assert len(min_gwds_similarity) == len(min_gwds_preference) == len(matching_rate_similarity) == len(matching_rate_preference), \
    "Loaded arrays must have the same length."

fig, ax = plt.subplots(figsize=(26, 6))  # Wide figure to fit 100*2 labels

# Plot lines + markers
line_sim, = ax.plot(iterations, min_gwds_similarity, marker='o', linewidth=1.5, label="Similarity", alpha=0.9)
line_pref, = ax.plot(iterations, min_gwds_preference, marker='s', linewidth=1.5, label="Preference", alpha=0.9)

# Colors to match the plotted lines
color_sim = line_sim.get_color()
color_pref = line_pref.get_color()

# Vertical offsets for text labels, scaled to data range so labels don't overlap points
ymin = float(np.nanmin([min_gwds_similarity, min_gwds_preference]))
ymax = float(np.nanmax([min_gwds_similarity, min_gwds_preference]))
dy_up = 0.015 * (ymax - ymin) if ymax > ymin else 0.01
dy_dn = 0.015 * (ymax - ymin) if ymax > ymin else 0.01

# Annotate best matching rates at each iteration
for x, y_sim, y_pref, mr_sim, mr_pref in zip(
    iterations, min_gwds_similarity, min_gwds_preference, matching_rate_similarity, matching_rate_preference
):
    # Above similarity point
    ax.text(x, y_sim + dy_up, f"{mr_sim:.2f}", ha='center', va='bottom', fontsize=7, rotation=90, color=color_sim)
    # Below preference point
    ax.text(x, y_pref - dy_dn, f"{mr_pref:.2f}", ha='center', va='top', fontsize=7, rotation=90, color=color_pref)

# Cosmetics
ax.set_xlabel("Iteration")
ax.set_ylabel("Minimal GWD")
ax.set_title("Minimal GWDs Across Iterations\n(Labels show best matching rates at the epsilon of minimal GWD)")
ax.grid(True, linestyle='--', alpha=0.5)
ax.legend(loc="upper right")

# X ticks every iteration (100), but you can thin them if crowded:
ax.set_xticks(iterations)
# ax.set_xticks(iterations[::5])  # <- uncomment to show every 5th tick instead

plt.tight_layout()
# plt.savefig("minimal_gwds_with_best_matching_rates.png", dpi=300)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import random

# Example: choose which set to plot — here I use preference
OT_plans = OT_plan_preference  # or OT_plan_similarity

# ---------------- 1) Random selection of 10 OT plans ----------------
n_to_plot = 10
if len(OT_plans) < n_to_plot:
    raise ValueError(f"Not enough OT plans. Found {len(OT_plans)}.")

random_indices = random.sample(range(len(OT_plans)), n_to_plot)
sampled_plans = [OT_plans[i] for i in random_indices]

ncols = 5
nrows = int(np.ceil(n_to_plot / ncols))

fig, axes = plt.subplots(nrows, ncols, figsize=(4*ncols, 4*nrows))
axes = axes.flatten()

for ax, idx, plan in zip(axes, random_indices, sampled_plans):
    im = ax.imshow(plan, cmap="viridis", aspect="auto")
    ax.set_title(f"OT Plan #{idx}")


# Remove any unused subplots if n_to_plot isn't multiple of ncols
for ax in axes[len(sampled_plans):]:
    ax.axis("off")

fig.colorbar(im, ax=axes, orientation="vertical", fraction=0.02, pad=0.04)
plt.suptitle("Random 10 OT Plans", fontsize=16)
# plt.tight_layout()
plt.show()

# ---------------- 2) Average OT plan across all iterations ----------------
avg_plan = np.mean(np.stack(OT_plans, axis=0), axis=0)

plt.figure(figsize=(6, 5))
plt.imshow(avg_plan, cmap="viridis", aspect="auto")
plt.colorbar(label="Average Transport Mass")
plt.title(f"Average OT Plan (n={len(OT_plans)} iterations)")
plt.xlabel("Target Space")
plt.ylabel("Source Space")
plt.tight_layout()
plt.show()
```
