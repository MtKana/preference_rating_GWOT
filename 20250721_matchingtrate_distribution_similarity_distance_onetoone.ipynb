```python
from utilityFunctions import load_csv_to_matrix_batch, compute_color_preference_distance_batch, show_heatmaps, compute_color_similarity_distance_batch, compute_correlations, RSA, split_and_average_matrices
from utilityFunctions_GWOT import compute_GWOT_for_all_pairs, GWD_and_plot
import numpy as np
import pickle
import os
import random
import pandas as pd
import matplotlib.pyplot as plt

unique_colours = np.array(['#d2b700', '#db8b08', '#c7512c', '#c13547', '#a03663', '#753a7a', '#4b488e', '#005692', '#006a8b', '#007b75', '#008a52', '#9aa400'])
colour_index = {colour: idx for idx, colour in enumerate(unique_colours)}
matrix_size = len(unique_colours)
response_type = 'similarity'
```

```python
folder_path = 'raw_data/202411_data'
response_type = 'similarity'
subject_matrices_similarity_raw = load_csv_to_matrix_batch(folder_path, response_type, colour_index, matrix_size)
subject_matrices_similarity_distance = compute_color_similarity_distance_batch(subject_matrices_similarity_raw)
```

```python
# --- Configuration ---
n_iterations = 20
results_dir = "gwot_results_similarity_distance_random_individuals"
os.makedirs(results_dir, exist_ok=True)

n_eps = 15
eps_range = [0.04, 5]
epsilons = np.logspace(np.log10(eps_range[0]), np.log10(eps_range[1]), n_eps)

# --- Required: your data list ---
# subject_matrices_similarity_distance = [...] # <-- must be defined

# --- Functions ---
def save_result(index, data):
    with open(os.path.join(results_dir, f"iteration_{index:03}.pkl"), "wb") as f:
        pickle.dump(data, f)

def get_completed_iterations():
    return {
        int(f.split("_")[1].split(".")[0])
        for f in os.listdir(results_dir)
        if f.startswith("iteration_") and f.endswith(".pkl")
    }

# --- Run 100 GWOTs between random pairs ---
completed = get_completed_iterations()
n_subjects = len(subject_matrices_similarity_distance)

for iteration in range(n_iterations):
    if iteration in completed:
        print(f"Iteration {iteration} already done. Skipping.")
        continue

    print(f"\nRunning iteration {iteration}...")

    # Select two different random indices
    idx1, idx2 = random.sample(range(n_subjects), 2)
    mat1 = subject_matrices_similarity_distance[idx1]
    mat2 = subject_matrices_similarity_distance[idx2]

    # Run GWOT
    pair_of_preference = [mat1, mat2]
    OT_plan, gwds, matching_rates, min_gwd, best_matching_rate = GWD_and_plot(
        pair_of_preference[0], pair_of_preference[1], epsilons
    )

    # Save results
    result_data = {
        "iteration": iteration,
        "subject1_index": idx1,
        "subject2_index": idx2,
        "OT_plan": OT_plan,
        "gwds": gwds,
        "matching_rates": matching_rates,
        "min_gwd": min_gwd,
        "best_matching_rate": best_matching_rate
    }
    save_result(iteration, result_data)

    print(f"Iteration {iteration} saved.")
```

```python
# --- Load results ---
matching_rates = []
iteration_ids = []

for fname in sorted(os.listdir(results_dir)):
    if fname.startswith("iteration_") and fname.endswith(".pkl"):
        with open(os.path.join(results_dir, fname), "rb") as f:
            data = pickle.load(f)
            matching_rates.append(data["best_matching_rate"])
            iteration_ids.append(data["iteration"])

# --- Horizontal Bar Graph (Histogram) ---
plt.figure(figsize=(10, 6))
plt.hist(matching_rates, bins=20, orientation='horizontal', color='skyblue', edgecolor='black')
plt.title("Distribution of Best Matching Rates (Random Individuals)")
plt.xlabel("Number of Iterations")
plt.ylabel("Best Matching Rate")
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()

# --- Scatter Plot with Matching Rates ---
plt.figure(figsize=(12, 5))
plt.plot(iteration_ids, matching_rates, marker='o', linestyle='-', color='purple')

# Annotate each point
for x, y in zip(iteration_ids, matching_rates):
    plt.text(x, y + 0.005, f"{y:.3f}", ha='center', va='bottom', fontsize=8, rotation=45)

plt.title("Best Matching Rate per Iteration (Random Individuals)")
plt.xlabel("Iteration")
plt.ylabel("Best Matching Rate")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
from scipy.stats import mannwhitneyu


def load_matching_rates(folder):
    rates = []
    for fname in sorted(os.listdir(folder)):
        if fname.startswith("iteration_") and fname.endswith(".pkl"):
            with open(os.path.join(folder, fname), "rb") as f:
                data = pickle.load(f)
                if "best_matching_rate" in data:
                    rates.append(data["best_matching_rate"])
    return rates

# Load data
group_folder = "gwot_results_similarity_distance_averaged"
individual_folder = "gwot_results_similarity_distance_random_individuals"

group_rates = load_matching_rates(group_folder)
individual_rates = load_matching_rates(individual_folder)

# Mann–Whitney U test
stat, p_val = mannwhitneyu(group_rates, individual_rates, alternative='two-sided')

# Compute summary statistics
mean_group = np.mean(group_rates)
median_group = np.median(group_rates)
mean_ind = np.mean(individual_rates)
median_ind = np.median(individual_rates)

# --- Output ---
print("== Group vs. Individual Matching Rates ==")
print(f"Group-Averaged   Mean:   {mean_group:.4f}")
print(f"Group-Averaged   Median: {median_group:.4f}")
print(f"Individual Align Mean:   {mean_ind:.4f}")
print(f"Individual Align Median: {median_ind:.4f}")

print(f"\nMann–Whitney U statistic: {stat:.4f}")
print(f"P-value:                  {p_val:.4e}")

# Significance check
alpha = 0.05
if p_val < alpha:
    print("\n✅ The result is statistically significant (p < 0.05).")
else:
    print("\n❌ The result is NOT statistically significant (p ≥ 0.05).")
```

```python
def load_ot_plans(folder):
    ot_plans = []
    for fname in sorted(os.listdir(folder)):
        if fname.startswith("iteration_") and fname.endswith(".pkl"):
            with open(os.path.join(folder, fname), "rb") as f:
                data = pickle.load(f)
                if "OT_plan" in data:
                    ot_plans.append(data["OT_plan"])
    return np.array(ot_plans)

group_ot = load_ot_plans(group_folder)
individual_ot = load_ot_plans(individual_folder)

# --- Compute average OT plans ---
avg_ot_group = np.mean(group_ot, axis=0)
avg_ot_individual = np.mean(individual_ot, axis=0)

# --- Plot heatmaps ---
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

im1 = axes[0].imshow(avg_ot_group, cmap='viridis')
axes[0].set_title("Avg OT Plan – Group-Averaged")
axes[0].set_xlabel("Target Index")
axes[0].set_ylabel("Source Index")
plt.colorbar(im1, ax=axes[0], fraction=0.046, pad=0.04)

im2 = axes[1].imshow(avg_ot_individual, cmap='viridis')
axes[1].set_title("Avg OT Plan – Individual-Based")
axes[1].set_xlabel("Target Index")
axes[1].set_ylabel("Source Index")
plt.colorbar(im2, ax=axes[1], fraction=0.046, pad=0.04)

plt.tight_layout()
plt.show()
```
