```python
import numpy as np
import random
import matplotlib.pyplot as plt
from utilityFunctions import (
    compute_color_preference_distance_batch,
    load_csv_to_matrix_batch,
    compute_color_similarity_distance_batch,
)
from utilityFunctions_GWOT import GWD_and_plot
import os

# Configuration
n_iterations = 10
n_subjects = 19
folder_path = 'raw_data/202411_data'
response_type = 'preference'
unique_colours = np.array(['#d2b700', '#db8b08', '#c7512c', '#c13547', '#a03663', '#753a7a',
                           '#4b488e', '#005692', '#006a8b', '#007b75', '#008a52', '#9aa400'])
colour_index = {colour: idx for idx, colour in enumerate(unique_colours)}
matrix_size = len(unique_colours)

# Load raw preference matrices
subject_matrices_preference_raw = load_csv_to_matrix_batch(
    folder_path,
    response_type,
    colour_index,
    matrix_size
)

subject_matrices_preference_distance = compute_color_preference_distance_batch(subject_matrices_preference_raw)

# Function to reorder preference matrices
def reorder_preference_matrix(matrix):
    row_means = np.mean(matrix, axis=1)
    sorted_indices = np.argsort(row_means)[::-1]  # Sort rows by descending average value
    reordered_matrix = matrix[sorted_indices, :][:, sorted_indices]  # Reorder rows and columns
    return reordered_matrix

# Function to shuffle rows of a matrix
def shuffle_matrix_rows(matrix):
    shuffled_matrix = matrix.copy()
    for i in range(matrix.shape[0]):
        np.random.shuffle(shuffled_matrix[i])
    return shuffled_matrix

# Function to perform GWOT on randomly split groups
def perform_gwot(subject_matrices, n_iterations):
    min_gwds = []
    for iteration in range(n_iterations):
        indices = list(range(len(subject_matrices)))
        random.shuffle(indices)
        group1_idx = indices[:len(indices) // 2]
        group2_idx = indices[len(indices) // 2:]

        group1_matrices = [subject_matrices[i] for i in group1_idx]
        group2_matrices = [subject_matrices[i] for i in group2_idx]

        avg_group1 = np.mean(group1_matrices, axis=0)
        avg_group2 = np.mean(group2_matrices, axis=0)

        _, _, _, min_gwd, _ = GWD_and_plot(avg_group1, avg_group2, epsilons=np.logspace(-2, 1, 25))
        min_gwds.append(min_gwd)
    return min_gwds

# Step 1: Reorder preference matrices
reordered_matrices = [reorder_preference_matrix(matrix) for matrix in subject_matrices_preference_distance]

# Step 2: Perform GWOT on reordered matrices
min_gwds_reordered = perform_gwot(reordered_matrices, n_iterations)

# Step 3: Shuffle reordered matrices and perform GWOT
shuffled_matrices = [shuffle_matrix_rows(matrix) for matrix in reordered_matrices]
min_gwds_shuffled = perform_gwot(shuffled_matrices, n_iterations)

# Step 4: Plot distributions of minimal GWD values
plt.figure(figsize=(10, 6))
plt.hist(min_gwds_reordered, bins=10, alpha=0.7, label="Reordered Matrices", color="blue")
plt.hist(min_gwds_shuffled, bins=10, alpha=0.7, label="Shuffled Matrices", color="orange")
plt.xlabel("Minimal GWD")
plt.ylabel("Frequency")
plt.title("Distribution of Minimal GWD Values")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()
```
